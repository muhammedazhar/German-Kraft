```python
# Loaded variable 'df' from URI: /Users/muhammedazhar/Developer/German-Kraft/Datasets/Sales by Product and Modifier_cleaned.csv
import pandas as pd
df = pd.read_csv(r'/Users/muhammedazhar/Developer/German-Kraft/Datasets/Sales by Product and Modifier_cleaned.csv')
```

```python
# Display unique values for each column with better formatting
print("=" * 80)
print("UNIQUE VALUES ANALYSIS")
print("=" * 80)

print("\n📊 CATEGORIES")
print("-" * 80)
for i, cat in enumerate(df['Category'].unique(), 1):
    print(f"  {i}. {cat}")
print(f"\n  ➤ Total unique categories: {df['Category'].nunique()}")

print("\n" + "=" * 80)
print("\n🔧 MODIFIERS")
print("-" * 80)
modifiers = df['Modifiers'].dropna().unique()
for i, mod in enumerate(modifiers, 1):
    print(f"  {i}. {mod}")
print(f"\n  ➤ Total unique modifiers: {df['Modifiers'].nunique()}")

print("\n" + "=" * 80)
print("\n🍺 PRODUCTS")
print("-" * 80)
for i, prod in enumerate(df['Product'].unique(), 1):
    print(f"  {i}. {prod}")
print(f"\n  ➤ Total unique products: {df['Product'].nunique()}")

print("\n" + "=" * 80)
print("\n🥤 MIXERS")
print("-" * 80)
mixers = df['Mixer'].dropna().unique()
for i, mix in enumerate(mixers, 1):
    print(f"  {i}. {mix}")
print(f"\n  ➤ Total unique mixers: {df['Mixer'].nunique()}")
print("\n" + "=" * 80)
```
